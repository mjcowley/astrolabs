# Part 1: Measuring the Universe

Distances in astronomy are so large that kilometres quickly become useless. Astronomers use a ladder of units, each suited to a different scale:

| Unit | Symbol | Definition | Used for |
| --- | :-: | --- | --- |
| kilometre | km | 1,000 metres | The Earth and Moon |
| astronomical unit | AU | Average Earth to Sun distance, about 150 million km | The Solar System |
| light-year | ly | Distance light travels in one year, about 9.46 trillion km | Stars and galaxies |
| parsec | pc | Distance at which a star shifts by 1 arcsecond as the Earth orbits the Sun, about 3.26 ly | Professional astronomy, from stars to the whole Universe |

Cosmologists usually work in **megaparsecs** (Mpc): one million parsecs.

## Python as a calculator

Python can do arithmetic just like a calculator. ▶️ Run the cell below.

```python
# How many kilometres in one light-year?
seconds_in_a_year = 365.25 * 24 * 60 * 60
light_year_km = 299792.458 * seconds_in_a_year

print("Seconds in a year:", seconds_in_a_year)
print("One light-year in km:", light_year_km)
```

That second number is printed in **scientific notation**: `9.46e+12` means $9.46 \times 10^{12}$, or about 9.46 trillion km.

## Letting Python handle the units

Keeping track of units by hand is where most mistakes happen. The `astropy.units` package attaches units to numbers and converts between them for us. We write a number, then multiply by a unit like `u.km`, and use `.to()` to convert.

```python
# 👉 Try changing the distance or the units!
distance_to_moon = 384400 * u.km

print(distance_to_moon)
print(distance_to_moon.to(u.AU))
print(distance_to_moon.to(u.lyr))
```

## Looking back in time

Light travels incredibly fast, about 300,000 km every second, but not infinitely fast. When we look at anything in the sky we see it as it **was** when the light left it, not as it is now. This is called **lookback time**.

The time light takes to reach us is just distance divided by speed: $t = d / c$.

```python
distance_to_sun = 1 * u.AU

light_travel_time = distance_to_sun / c
print("Light from the Sun takes:", light_travel_time.to(u.s))
print("That's:", light_travel_time.to(u.min))
```

# Part 2: Powers of Ten

In the lecture we zoomed out from the Earth and Moon to the whole Universe. Let's put numbers on that zoom.

## Working with many numbers at once

So far we've worked with one number at a time. A **NumPy array** holds a whole list of numbers, and lets us do maths on all of them in one go.

```python
# Pull the names and distances out of our dictionary
names = list(cosmic_objects.keys())
distances_km = u.Quantity([d.to(u.km) for d in cosmic_objects.values()])

print(distances_km)
```

```python
# One line converts every distance at once
print(distances_km.to(u.lyr))
```

These numbers span a ridiculous range. To compare them, astronomers use **powers of ten**. The function `np.log10()` tells us the power of ten of a number: `np.log10(1000)` is 3 because $1000 = 10^3$.

```python
powers_of_ten = np.log10(distances_km.value)

for name, power in zip(names, powers_of_ten):
    print(f"{name:38s} 10^{power:.1f} km")
```

> 🔍 **Peek inside:** `for name, power in zip(...)` is a **loop**. It repeats the indented line once for each object. The `f"..."` is an **f-string**, a handy way to put numbers inside text.

## Plotting the cosmic ladder

Now let's **plot** it. Look at how the plot changes when we switch to a **logarithmic** axis, where each step is ten times bigger than the last.

```python
# 👉 Change log_axis to False to see a normal (linear) axis
log_axis = True

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(names, distances_km.value, color=al.COLOURS["blue"])
if log_axis:
    ax.set_xscale("log")
ax.set_xlabel("Distance (km)")
ax.set_title("The cosmic distance ladder")
ax.invert_yaxis()
plt.show()
```

With a linear axis, everything except MoM-z14 is squashed to zero. The log axis is the only way to see them all, which is why you'll see log axes all through this unit.

> 🎯 **Canvas checkpoint Q4:** How many powers of ten (orders of magnitude) separate the Moon from MoM-z14? Use the cell below.

```python
# 👉 Run this, then round the answer to one decimal place
ratio = cosmic_objects["MoM-z14 (most distant known galaxy)"] / cosmic_objects["The Moon"]
orders_of_magnitude = np.log10(ratio.decompose().value)
print(f"Orders of magnitude between the Moon and MoM-z14: {orders_of_magnitude:.1f}")

al.check("C7_orders", orders_of_magnitude)
```

# Part 3: Testing the Cosmological Principle

The **cosmological principle** says that on the largest scales the Universe is:

* **homogeneous:** the same everywhere (no special places), and
* **isotropic:** the same in every direction (no special directions).

On small scales, though, the Universe is clumpy. In this part you'll test both ideas with real data.

## Loading real data

Data usually comes as a **table**: rows for each object, columns for each measurement. The `pandas` package loads tables into a **DataFrame**.

We'll start with **gamma-ray bursts** (GRBs): the most powerful explosions in the Universe, thought to happen when massive stars collapse into black holes. These were detected by the BATSE instrument on NASA's Compton Gamma Ray Observatory between 1991 and 2000.

```python
grbs = pd.read_csv(al.get_data("C7/batse_grbs.csv"))

grbs.head()    # show the first five rows
```

```python
print("Number of GRBs:", len(grbs))
print("Columns:", list(grbs.columns))
```

Each GRB has a position on the sky. We'll use **galactic coordinates**: `glon` (galactic longitude) runs around the plane of the Milky Way, and `glat` (galactic latitude) measures how far above (+90°) or below (−90°) the plane something is.

## Making a map of the whole sky

To show the whole sky on a flat page we use the same trick as world maps: a **Mollweide projection**. The Milky Way runs along the middle.

```python
def sky_map(lon, lat, title, colour):
    # Convert degrees to the form matplotlib's sky maps expect
    lon_rad = np.radians(-((lon + 180) % 360 - 180))
    lat_rad = np.radians(lat)

    fig = plt.figure(figsize=(10, 5))
    ax = fig.add_subplot(projection="mollweide")
    ax.scatter(lon_rad, lat_rad, s=3, color=colour, alpha=0.6)
    ax.set_title(title, pad=20)
    ax.set_xticklabels([])
    plt.show()

sky_map(grbs["glon"], grbs["glat"], "Gamma-ray bursts (BATSE)", al.COLOURS["orange"])
```

> 🔍 **Peek inside:** `def sky_map(...)` defines our own **function**: a reusable recipe. Now we can draw any sky map with one line.

Does that look clumpy, or evenly spread? Let's test it with numbers. If GRBs are isotropic, about half should be above the plane of the Milky Way and half below.

```python
north = grbs["glat"] > 0          # True for each GRB above the plane, False below
fraction_north = north.mean()     # the average of True/False is the fraction that are True

print(f"GRBs above the Milky Way plane: {north.sum()} of {len(grbs)}")
print(f"Fraction above: {fraction_north:.3f}")
```

## Now try nearby galaxies

Let's compare that with galaxies from the **2MASS Redshift Survey**, a map of the nearest tens of thousands of galaxies. Each has a distance in megaparsecs (Mpc), estimated from how fast it's moving away from us (you'll learn how in Week 4). Remember, 1 Mpc is about 3.26 million light-years.

```python
galaxies = pd.read_csv(al.get_data("C7/nearby_galaxies.csv"))

print("Number of galaxies:", len(galaxies))
galaxies.head()
```

```python
# 👉 Change the maximum distance (in Mpc) and re-run
max_distance = 30

nearby = galaxies[galaxies["distance_mpc"] < max_distance]
sky_map(nearby["glon"], nearby["glat"],
        f"{len(nearby)} galaxies closer than {max_distance} Mpc", al.COLOURS["blue"])

fraction_north = (nearby["glat"] > 0).mean()
print(f"Fraction above the Milky Way plane: {fraction_north:.2f}")
```

The empty band across the middle is the **zone of avoidance**: dust in our own Milky Way blocks the view, so this survey can't see galaxies there. It's a problem with where we're standing, not with the Universe! Gamma rays shine straight through the dust, which is why the GRB map has no gap.

Nearby galaxies are clumped into groups, clusters and filaments. Now see what happens as you include more and more distant galaxies. ▶️ Run the cell below and move the slider.

```python
al.isotropy_explorer(galaxies, sky_map)
```

# Part 4: Why Is the Night Sky Dark?

In the lecture we met **Olbers' paradox**. If the Universe were infinite, eternal and unchanging, every line of sight would eventually end on the surface of a star, just like every line of sight in a big enough forest ends at a tree. The whole sky should be as bright as the surface of the Sun!

Let's test this with a **toy universe**: a patch of sky filled evenly with identical stars. The **age** sets how far away light can reach us from: in a universe of age $t$, we can only see stars closer than $d = c \times t$.

▶️ Run the cell below. Start with a young universe, then slowly increase the age.

```python
al.olbers_explorer()
```

> 🔍 **Peek inside:** the toy universe is the `olbers_sky` function in `astrolabs.py`. It scatters stars randomly through space and paints them onto the sky, far ones first, so nearer stars block the ones behind them.

> 🎯 **Canvas checkpoint Q8:** In a **static** universe, set the age to the maximum (200). Roughly what percentage of the sky is covered by stars?
>
> 🎯 **Canvas checkpoint Q9:** Keep the age at 200 and tick **Expanding universe**. The sky is still covered by stars, but what happens to its brightness, and why?

Your toy universe shows two of the resolutions from the lecture:

* **The Universe has a finite age:** light from stars beyond $d = c\,t_{\rm age}$ hasn't reached us yet.
* **The Universe is expanding:** light from distant stars is **redshifted**, stretching it to longer wavelengths and dimming it into darkness.

# Part 5: Sneak Peek: Your First Astronomical Image

Telescopes don't take photos like your phone. They record how much light lands on each pixel through one **filter** at a time, and save it as a **FITS** file (Flexible Image Transport System), the standard file format in astronomy.

An image is really just a big table of numbers: one brightness value per pixel. Let's open one of **NGC 1365**, the Great Barred Spiral Galaxy, about 56 million light-years away in the southern constellation Fornax. It was taken by the Dark Energy Camera in Chile for the DESI Legacy Imaging Surveys.

```python
from astropy.io import fits

hdul = fits.open(al.get_data("C7/galaxy_grz.fits"))
hdul.info()
```

```python
header = hdul[0].header
cube = hdul[0].data

print("Shape of the data:", cube.shape)
print("Filters:", header["BAND0"], header["BAND1"], header["BAND2"])
print("Object:", header["OBJECT"])
```

The data has **three layers**, one for each filter: **g** (green), **r** (red) and **z** (near-infrared). Each layer is a grid of pixels. Let's look at one layer on its own.

```python
g_band, r_band, z_band = cube[0], cube[1], cube[2]

print("Brightest pixel:", g_band.max())
print("Typical pixel:", np.median(g_band))
```

Notice the brightest pixel is far brighter than a typical one. That makes images hard to display, as you'll see. ▶️ Run the cell below. With a **linear** stretch the image looks almost empty! Try the other stretches and move the clip slider until the galaxy appears.

```python
al.image_explorer(r_band, title="r band")
```

## Adding colour

A colour image is made by combining three filters: the longest wavelength as red, the middle as green and the shortest as blue. ▶️ Run the cell below and adjust the sliders to make your best image.
